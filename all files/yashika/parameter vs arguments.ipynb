# parameter vs arguments

```python
def stu(name):
    print(name)
stu("Yashika")
```

# positional arguments

```python
def student(name,age):
    print("name:",name)
    print("age:",age)
student("yashika",18)
```

```python
def stu(age,name):
    print("age:",age)
    print("name:",name)
stu("yashika",18)
```

# / positional-only

```python
def stu(name,age, /):
    print(name,age)
stu("yashika",18)
```

# * keyword only

```python
def student(*,name,age):
    print(name,age)
```

```python
student(name="yashika",age=18)
```

# keyword arguments

```python
def student(name,age):
    print(name,age)
student(name="yashika",age=18)
```

# positional + keyword

```python
def student(name,age,city):
    print(name,age,city)
student(name="yashika",age=18,city="chandigarh")
```

# * args

```python
def add(*args):
    print(args)
add(10,20,50,90)
```

```python
def add(*b):
    print(b)
add(10,80,50,60,80)
```

```python
def add(*args):
    total=0
    for number in args:
        total+= number
        print(total)
add(10,20,30)
```

# **kwargs

```python
def stu(**kwargs):
    print(kwargs)
stu(name="yashika",age=18,city="chandigarh")
```
