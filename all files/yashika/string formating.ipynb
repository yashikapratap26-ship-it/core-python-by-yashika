# String formating

```python
name="yashika"
place="panchkula"
mail="hey my name is {} and i live in {}"
print(mail.format(name,place))
```

```python
print(mail.format(place,name))
```

# new format

```python
print(f"(hey my name is {name} and i live in {place}")
```

```python
print(f"hey my name is {{name}} and i live in {{place}}")
```

```python
t="my number {} so i will buy this product in {}"
print(t.format(5+6,11*3))
```

```python
print(f"my number{5+6} so i will buy this product in{11*3}")
```

```python
t="for only{price:.2f} dollars"
```

```python
print(t.format(price=98.678))
```
