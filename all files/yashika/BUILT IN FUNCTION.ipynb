```python
def area(length,widhth):
    print(length*widhth)
area(10,60)
```

```python
def area(radius):
    print(3.14*radius**2)
area(4)
```

```python
j=[10,20,30,40,50]
def average(n):
    total=0
    count=0
    for i in n:
        total=total+i
        count=count+1
        average=(total/count)
    print(average)
```

```python
average(j)
```

```python
def three_big(a,b,c):
    if a > b and a > c:
        print("a is bigger")
    elif b > a and c > b:
        print("b is greater")
    else:
        print("c is bigger")
```

```python
three_big(50,25,41)
```

```python
now=int(input("tell current year"))
birth=int(input("tell your year of birth"))
age(now,birth)
```
