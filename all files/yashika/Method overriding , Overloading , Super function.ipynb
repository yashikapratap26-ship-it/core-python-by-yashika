# 1) Method Overriding

# Meaning = A child class provides it own version of a method that olready exists in the parent class.

```python
class Father:
    def work(self):
        print("Father is working")


class Son(Father):
    def work(self):
        print("son is working")

s = Son()
s.work()
```

# 2) Method Overloading

# Meaning = using the same method name with different numbers/types of arguments.

```python
class Calculator:
    def add(self, a, b):
        return a + b

    def add(self,a, b, c):
        return a + b + c
```

```python
c = Calculator()
c.add(5,8,9)
```

# The second add() replaces the first one.

```python
class Calculator:

    def add(self, a, b, c=0):
        return a + b + c

c = Calculator()

print(c.add(10,20))
print(c.add(10,20,50))
```

# Super function

```python
class Father:
    def work(self):
        print("Father is working")

class Son(Father):
    def work(self):
        super().work()
        print("Son is working")

s = Son()
s.work()
```

# Super() with init()

```python
class Father:
    def __init__(self):
        print("Father constructor")

class Son(Father):
    def __init__(self):
        super().__init__()
        print("Son constructor")

s = Son()
        
```
