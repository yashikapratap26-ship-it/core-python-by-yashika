# 2) multiple

```python
class Person:
    def person_info(self,name,age):
        print('Inside person class')
        print('Name:', name, 'Age:', age)

class Company:
    def company_info(self,company_name,location):
        print('Inside company class')
        print('Name:', company_name, 'location:', location)

class Employee(Company,Person):
    def Employee_info(self,salary,skill):
        print('Inside employee class')
        print('Salary:', salary, 'Skill:', skill)

emp = Employee()
emp.company_info("tech","chd")
emp.person_info("sal",43)
```

# 3) multilevel

```python
class Vehicle:
    def Vehicle_info(self):
        print('Inside Vehicle class')

class Vehicle1(Vehicle):
    def Vehicle_info1(self):
           print('Inside Vehicle class')

class Car(Vehicle1):
    def Car_info1(self):
           print('Inside Car class')

class SportsCar(Car):
    def Sports_car_info1(self):
           print('Inside SportsCar class')

s_car = SportsCar()
s_car.Vehicle_info1()
s_car.Car_info1()
s_car.Sports_car_info1()
```

# Hierarchical

```python
class Vehicle:
    def info(self):
        print("This is Vehicle")

class Car(Vehicle):
    def car_info(self,name):
        print("Car name is:" , name)

class Truck(Vehicle):
    def truck_info(self,name):
        print("Truck name is:" , name)

obj1 = Car()
obj1.info()
obj1.car_info('BMW')

obj2 = Truck()
obj2.info()
obj2.truck_info('ford')
```

```python
class A:
    def method_a(self):
        print("A")

class B:
    def method_b(self):
        print("B")

class C(A):
    def method_c(self):
        print("C")

class D(B,C):
    def method_d(self):
        print("D")

d = D()
d.method_a()
d.method_b()
d.method_c()
d.method_d()
```
