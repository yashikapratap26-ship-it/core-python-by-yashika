# fizz buzz

```python

for i in range(0,100):
    if i %3==0 and i % 5==0:
        print("fizzbuzz")
    elif i % 5==0:
        print("buzz")
    elif i %3==0:
        print("fizz")
    else:
        print(i)
      
```

# average of numbers from 1 to 10

```python
total=0
count=0
for i in range(1,11):
    total = total + i
    count = count + 1
average = total / count
print(average)
```
