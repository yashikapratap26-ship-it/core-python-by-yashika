# Public

# In python, normally we simply write the variable name.

```python
class Student:
    def __init__(self):
        self.name = "Rahul"             # Public

s1 = Student()

print(s1.name)
```

```python

s1.name = "Yashika"
print(s1.name)
```

# Private

# Private members are written with two underscores_.

```python
class bank:
    def __init__(self):
        self.__balance = 10000

b = bank()

print(b.__balance)
```

# We can acess the data by adding other user define function on it.

```python
class bank:
    def __init__(self):
        self.__balance = 10000

    def show_balance(self):
        print(self.__balance)


b = bank()

b.show_balance()
```

# Protected

# Procted behaves more like public in python because you technically can access it from outside,but the_tells programmers:"this is intended for internal/subclass use."

```python
class student:
    def __init__(self):
        self._marks = 80

s1 = student()

print(s1._marks)
```

# Best easy example

```python
class student:
    def __init__(self):
        self.name = "Yashika"          # Public
        self._age = 18                # Protected
        self.__marks = 90             # Private

s = student()

print(s.name)        #                     
print(s._age)        # wWorks in python
print(s.__marks)     # Error
```
